```python
# --- Imports ---
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import os

# --- Paths ---
MODEL_PATH = r"train2/weights/best.pt"   # your YOLOv11 trained weights
TEST_IMAGES = ["fire.jpg", "knife.jpg", "cctv2.jpeg"]  # images in current directory

# --- Load YOLOv11 model ---
print("🚀 Loading YOLOv11 model...")
model = YOLO(MODEL_PATH)

# --- Run detection on test images ---
os.makedirs("inference_results", exist_ok=True)

for img_name in TEST_IMAGES:
    img_path = img_name  # images are in the same directory as notebook
    if not os.path.exists(img_path):
        print(f"⚠️ File not found: {img_path}")
        continue
    
    print(f"\n🔍 Detecting objects in: {img_name}")
    results = model.predict(
        source=img_path,
        conf=0.4,        # confidence threshold
        iou=0.5,         # IOU threshold
        save=True,       # save results in runs/detect
        project="inference_results", 
        name="detectifai_v11"
    )

    # --- Display detection result in memory ---
    img = results[0].plot()  # Get annotated image as numpy array
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(8, 6))
    plt.imshow(img)
    plt.title(f"Detections for {img_name}")
    plt.axis("off")
    plt.show()

print("\n✅ Inference complete! Check 'inference_results/detectifai_v11' for output images.")
```
